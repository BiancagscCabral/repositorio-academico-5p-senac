```python
import re
import unicodedata

def remover_acentos(texto):
    # Função para remover acentos e facilitar o reconhecimento das palavras
    return ''.join(c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn')

def chatbot_responder(pergunta):
    # Normaliza a pergunta (deixa tudo minúsculo e remove acentos)
    pergunta_normalizada = remover_acentos(pergunta.lower())

    # Pergunta 1
    if "cnpj" in pergunta_normalizada:
        return "Sim. Hoje o cadastro exige um CNPJ válido (14 dígitos) é um campo obrigatório no formulário."

    # Pergunta 2
    elif "pago" in pergunta_normalizada or "cobrado" in pergunta_normalizada or "pagar" in pergunta_normalizada:
        return "Não. O cadastro é gratuito e seu perfil aparece normalmente nas buscas, sem nenhuma cobrança."

    # Pergunta 3
    elif "recomendacao" in pergunta_normalizada or "recomenda" in pergunta_normalizada:
        return "Quando uma empresa cadastra uma necessidade, comparamos com o que está no seu perfil serviços, tecnologias e segmentos e sua startup pode aparecer nas recomendações dela."

    # Pergunta 4
    elif "matchmaking" in pergunta_normalizada or "compatibilidade" in pergunta_normalizada:
        return "Comparamos os serviços e tecnologias que você pediu com o que cada startup entrega, o segmento em que ela já atua e as palavras-chave da descrição da sua necessidade. Cada recomendação vem com uma justificativa clara e nenhuma startup recebe nota negativa."

    # Pergunta 5
    elif "criar conta" in pergunta_normalizada or "pesquisar startups" in pergunta_normalizada or "conta para pesquisar" in pergunta_normalizada:
        return "Não. A busca e os filtros avançados por tecnologias, serviços e segmentos são abertos para qualquer visitante. O cadastro gratuito é necessário para visualizar os nomes e perfis completos das startups."

    # Pergunta 6
    elif "exige cadastro" in pergunta_normalizada or "precisa de cadastro" in pergunta_normalizada:
        return "Visualizar os dados e perfis completos das startups, salvar favoritas, cadastrar necessidades e solicitar matchmaking."

    # Resposta Padrão
    else:
        return ("Desculpe, não entendi sua pergunta. Você pode me perguntar sobre:\n"
                "- Se precisa de CNPJ\n"
                "- Se o cadastro é pago\n"
                "- Como funciona a recomendação ou o matchmaking\n"
                "- Se precisa de conta para pesquisar\n"
                "- O que exige cadastro")

# Função principal que roda o loop de interação do chat no Colab
def iniciar_chat():
    print("="*50)
    print("🤖 Chatbot da Plataforma de Startups inciado!")
    print("Equipe: Eduardo, Bianca, Gislany, João Marcos, Pedro e Reideclildon.")
    print("Digite 'sair' a qualquer momento para encerrar.")
    print("="*50)

    while True:
        entrada_usuario = input("\nVocê: ")

        # Condição de parada
        if entrada_usuario.lower().strip() == 'sair':
            print("Chatbot: Até logo! Encerrando o sistema...")
            break

        resposta = chatbot_responder(entrada_usuario)
        print(f"Chatbot: {resposta}")

# Inicia o chatbot
iniciar_chat()
```

```python
import unicodedata

# 1. Base de Dados de Perguntas e Respostas
# Aqui organizamos o conteúdo do documento original pelo número da pergunta
base_dados = {
    1: {
        "pergunta": "Preciso ter CNPJ pra me cadastrar?",
        "resposta": "Sim. Hoje o cadastro exige um CNPJ válido (14 dígitos) é um campo obrigatório no formulário."
    },
    2: {
        "pergunta": "O cadastro é pago?",
        "resposta": "Não. O cadastro é gratuito e seu perfil aparece normalmente nas buscas, sem nenhuma cobrança."
    },
    3: {
        "pergunta": "Como funciona a recomendação pra empresas?",
        "resposta": "Quando uma empresa cadastra uma necessidade, comparamos com o que está no seu perfil serviços, tecnologias e segmentos e sua startup pode aparecer nas recomendações dela."
    },
    4: {
        "pergunta": "Como o matchmaking avalia a compatibilidade?",
        "resposta": "Comparamos os serviços e tecnologias que você pediu com o que cada startup entrega, o segmento em que ela já atua e as palavras-chave da descrição da sua necessidade. Cada recomendação vem com uma justificativa clara e nenhuma startup recebe nota negativa."
    },
    5: {
        "pergunta": "Preciso criar conta para pesquisar startups?",
        "resposta": "Não. A busca e os filtros avançados por tecnologias, serviços e segmentos são abertos para qualquer visitante. O cadastro gratuito é necessário para visualizar os nomes e perfis completos das startups."
    },
    6: {
        "pergunta": "O que exige cadastro?",
        "resposta": "Visualizar os dados e perfis completos das startups, salvar favoritas, cadastrar necessidades e solicitar matchmaking."
    }
}

# 2. Mapeamento de Palavras-chave (O agrupamento de escopo)
# Relacionamos cada palavra-chave aos IDs das perguntas correspondentes do banco de dados acima
mapa_palavras_chave = {
    "cnpj": [1],
    "cadastro": [1, 2, 5, 6],
    "cadastrar": [1, 2, 5, 6],
    "pago": [2],
    "pagar": [2],
    "gratis": [2],
    "gratuito": [2],
    "recomendacao": [3, 4],
    "recomenda": [3, 4],
    "empresa": [3],
    "empresas": [3],
    "matchmaking": [4, 6],
    "match": [4],
    "compatibilidade": [4],
    "conta": [5],
    "pesquisar": [5],
    "pesquisa": [5],
    "busca": [5],
    "startup": [3, 4, 5, 6],
    "startups": [3, 4, 5, 6]
}

def remover_acentos(texto):
    return ''.join(c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn')

def buscar_ids_por_palavras(entrada):
    # Quebra a frase digitada em várias palavras
    entrada_normalizada = remover_acentos(entrada.lower()).split()
    ids_encontrados = set() # O 'set' evita que a mesma pergunta apareça duplicada na lista

    # Verifica cada palavra digitada pelo usuário e cruza com o nosso mapa de escopos
    for palavra in entrada_normalizada:
        for chave, ids in mapa_palavras_chave.items():
            if chave in palavra:
                ids_encontrados.update(ids)

    # Retorna uma lista ordenada com os números das perguntas encontradas
    return sorted(list(ids_encontrados))

def iniciar_chat():
    print("="*60)
    print("🤖 Chatbot da Plataforma de Startups (Com Opções)")
    print("Digite um tema (ex: 'cadastro', 'cnpj', 'matchmaking')")
    print("Digite 'sair' a qualquer momento para encerrar o programa.")
    print("="*60)

    estado = "pesquisar_tema"
    opcoes_atuais = []

    while True:

        # ESTADO 1: O Bot espera o usuário digitar uma palavra-chave / tema
        if estado == "pesquisar_tema":
            entrada_usuario = input("\nVocê (digite um assunto): ")

            if entrada_usuario.lower().strip() == 'sair':
                print("Chatbot: Até logo! Encerrando...")
                break

            ids_encontrados = buscar_ids_por_palavras(entrada_usuario)

            if not ids_encontrados:
                print("Chatbot: Não encontrei perguntas sobre isso. Tente palavras como: 'cadastro', 'cnpj', 'pago', 'matchmaking', 'pesquisar'.")
            else:
                print("\nChatbot: Encontrei as seguintes opções sobre esse tema. Escolha uma:")
                for i, q_id in enumerate(ids_encontrados):
                    print(f"[{i + 1}] {base_dados[q_id]['pergunta']}")

                print("[0] Voltar e pesquisar outro assunto")

                # Salva as opções mostradas e muda o estado para aguardar o número escolhido
                opcoes_atuais = ids_encontrados
                estado = "escolher_numero"

        # ESTADO 2: O Bot espera o usuário digitar o número da pergunta que ele quer ler
        elif estado == "escolher_numero":
            escolha = input("\nSua Escolha (digite o número): ").strip()

            if escolha.lower() == 'sair':
                print("Chatbot: Até logo! Encerrando...")
                break

            if escolha.isdigit():
                escolha_num = int(escolha)

                if escolha_num == 0: # Opção de voltar
                    estado = "pesquisar_tema"

                # Se a escolha for válida dentro das opções listadas
                elif 1 <= escolha_num <= len(opcoes_atuais):
                    # Acha o ID real da pergunta correspondente no banco de dados
                    q_id = opcoes_atuais[escolha_num - 1]

                    print(f"\n✅ RESPOSTA:\n{base_dados[q_id]['resposta']}\n")
                    print("-" * 60)

                    # Após responder, volta ao estágio inicial
                    estado = "pesquisar_tema"
                else:
                    print("Chatbot: Opção inválida. Digite um dos números listados.")
            else:
                print("Chatbot: Por favor, digite apenas números válidos.")

# Inicia o chatbot
iniciar_chat()
```
